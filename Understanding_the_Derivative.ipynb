# Exploring the Derivative as a Limit

This notebook provides an interactive way to explore the concept of the derivative of a function at a point defined as a limit:

image.png

Understanding this fundamental concept is crucial for deeper learning in calculus, particularly in understanding change and approximation techniques.


```python
# Import Necessary Libraries
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, Dropdown
```

## Function Definitions and Setup

In this section, we define several mathematical functions to explore their derivatives. Each function represents a different type of behavior:

- **Quadratic Function** \( f(x) = x^2 \): A basic polynomial function that curves upward. Its derivative increases linearly, reflecting the increasing slope as \( x \) increases.
- **Sine Function** \( f(x) = sin(x) \): A periodic function that oscillates between -1 and 1. Its derivative, \( cos(x) \), reflects the rate of change in the sine wave.
- **Exponential Function** \( f(x) = e^x \): An exponential growth function whose slope (derivative) also grows exponentially.


```python
# Function to Compute the Derivative
def derivative(f, a, h=0.01):
    return (f(a + h) - f(a)) / h

# Function Definitions
def f1(x): return x**2
def f2(x): return np.sin(x)
def f3(x): return np.exp(x)

# Dictionary of Functions
functions = {'x^2': f1, 'sin(x)': f2, 'e^x': f3}
```

## Interactive Visualization Function

The `plot_function` is a core part of our interactive exploration, allowing us to visualize how the derivative approximation changes as \( h \) approaches zero. Here's a step-by-step breakdown of what happens inside this function:

### Function Parameters
- `function_name`: A string that corresponds to the key in our `functions` dictionary, allowing the user to select which mathematical function to analyze.
- `a`: The point at which we want to explore the derivative. This can be any \( x \) value within the visible range of our plot.
- `h`: The increment that approaches zero, used to compute the derivative approximation through the difference quotient method.

### Core Components of the Function
1. **Function Retrieval**: `f = functions[function_name]` retrieves the function based on the selected name. This allows us to dynamically switch between different mathematical functions like x^2, sin(x), and  e^x for our derivative analysis.

2. **Derivative Approximation**:
   - `derivative_approx = derivative(f, a, h)` calculates the approximate derivative using the difference quotient:
image.png

   - This computation gives us a numerical approximation of the slope of the tangent line at point \( a \), based on the chosen \( h \).

3. **Data for Plotting**:
   - `x = np.linspace(a - 5, a + 5, 400)` creates an array of \( x \) values centered around \( a \) and spanning 10 units. This determines the range of the plot and ensures that the area around \( a \) is well-represented.
   - `y = f(x)` computes the function values at each point in \( x \), giving us the curve of the function.
   - `secant_y = f(a) + derivative_approx * (x - a)` calculates the \( y \)-values for the secant line, which approximates the tangent line when \( h \) is sufficiently small. This line is plotted to visually represent the derivative at point \( a \).

4. **Plotting**:
   - The function then sets up a plot with a specified size (`plt.figure(figsize=(10, 6))`).
   - `plt.plot(x, y, label=f'{function_name}')` plots the function's curve.
   - `plt.plot(x, secant_y, '--', label=f'Secant Line h={h:.2f}')` adds the secant line to the plot, using a dashed style to distinguish it from the function curve.
   - `plt.scatter([a], [f(a)], color='red')` marks the point \( a \) on the function, highlighting the specific point of interest where the derivative is being approximated.


### Visualization Outcome
This interactive visualization allows students to see how the slope of the secant line changes as \( h \) varies and how it approaches the slope of the tangent line as \( h \) gets smaller. By manipulating \( h \), students can practically explore the limit definition of the derivative and its implications on different types of functions.


```python
# Interactive Visualization Function
def plot_function(function_name, a, h):
    f = functions[function_name]
    derivative_approx = derivative(f, a, h)

    x = np.linspace(a - 5, a + 5, 400)
    y = f(x)
    secant_y = f(a) + derivative_approx * (x - a)

    plt.figure(figsize=(10, 6))
    plt.plot(x, y, label=f'{function_name}')
    plt.plot(x, secant_y, '--', label=f'Secant Line h={h:.2f}')
    plt.scatter([a], [f(a)], color='red') # Point of tangency
    plt.title(f"Function: {function_name} at x={a}")
    plt.legend()
    plt.grid(True)
    plt.show()
```

## Understanding the Graph and How to Use It

As you interact with the graph, there are several key insights and actions to focus on, which will deepen your understanding of calculus concepts, particularly the derivative as a limit.

### What You Should Understand from the Graph

1. **Function Curve**:
   - This is the main graph line that represents the selected function (e.g., \(x^2\), \(\sin(x)\), or \(e^x\)). Observing this curve helps you see how the function behaves over a range of values.

2. **Secant Line**:
   - The dashed line on the graph is the secant line, which connects two points on the function curve: \((a, f(a))\) and \((a + h, f(a + h))\). This line is crucial because its slope represents the derivative approximation when \( h \) is not zero.

3. **Point of Tangency** (Red Dot):
   - The red dot marks the point \(a\) on the function curve. This point is where you are approximating the derivative.

4. **Behavior as \( h \) Approaches Zero**:
   - Watch how the secant line changes as you adjust \( h \). As \( h \) gets smaller, the secant line should increasingly resemble the tangent line at \( a \) — indicating the derivative at that point. The slope of this line as \( h \) approaches zero is what you're calculating as the derivative.

### How to Use the Interactive Widget

- **Select a Function**: Use the dropdown menu to choose the function you want to explore. Each function behaves differently, and changing functions can show you how different types of functions are differentiated.

- **Adjust \( a \)**: Move the slider to change the point \( a \) where you want to approximate the derivative. This allows you to see how the derivative (slope of the tangent line) varies at different points along the curve.

- **Modify \( h \)**: This slider controls the value of \( h \), the small increment used to calculate the difference quotient. Decrease \( h \) to see how the secant line approaches the tangent line. This visual demonstration helps you understand the concept of the limit: as \( h \) gets very small, the difference quotient becomes a better approximation of the derivative.

This interactive exploration is designed to help you visually grasp the concept of limits and derivatives, which are fundamental in understanding change and motion in calculus.


```python
# Interactive Widgets
interact(plot_function,
         function_name=Dropdown(options=functions.keys(), value='x^2', description='Function:'),
         a=FloatSlider(min=-10, max=10, step=0.1, value=0, description='a:'),
         h=FloatSlider(min=0.001, max=1, step=0.001, value=0.01, description='h:'))
```
