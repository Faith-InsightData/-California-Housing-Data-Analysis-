Feature Engineering Plan:
Handling Class Imbalance:

Since Pulmonary Disease (1) cases are fewer than No Disease (0), we can use:

SMOTE (Synthetic Minority Oversampling Technique) to balance the classes.

Class weighting in machine learning models instead of oversampling.

Creating Interaction Features:
Risk Score: Combine Smoking, Pollution, and Family History into a composite risk factor.

Respiratory Symptoms Score: Sum of Breathing Issues, Chest Tightness, and Throat Discomfort.

Immune System Weakness Score: Combination of Immune Weakness and Stress Immune.
Scaling Continuous Variables:

Energy Level & Oxygen Saturation need scaling (MinMaxScaler or StandardScaler)

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import joblib  # To load the trained model

# Load the trained model
@st.cache_resource
def load_model():
    return joblib.load("random_forest_model.pkl")  # Ensure this file exists in the same directory

model = load_model()

# Streamlit App Title
st.title("Pulmonary Disease Prediction App 🚑")

st.markdown("""
This app predicts the likelihood of **Pulmonary Disease** based on patient data.
Simply enter the required details below and click **Predict**.
""")

# Feature names (Replace with actual column names from your dataset)
feature_names = ["Age", "Smoking_History", "BMI", "Air_Pollution_Level"]  # Replace with real features

# User input fields
user_input = {}
for feature in feature_names:
    user_input[feature] = st.number_input(f"Enter {feature}:", min_value=0.0, step=0.1)

# Convert input into a DataFrame
input_df = pd.DataFrame([user_input])

# Prediction button
if st.button("Predict Disease Risk"):
    prediction = model.predict(input_df)[0]  # Get prediction (0 or 1)
    probability = model.predict_proba(input_df)[0][1]  # Get probability score

    # Display result
    st.subheader("Prediction Result")
    if prediction == 1:
        st.error(f"⚠️ **High Risk of Pulmonary Disease!** (Confidence: {probability:.2%})")
    else:
        st.success(f"✅ **Low Risk of Pulmonary Disease** (Confidence: {probability:.2%})")

# Footer
st.write("📌 *Model trained using Scikit-Learn & deployed with Streamlit*")
```
